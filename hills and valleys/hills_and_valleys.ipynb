```python
import torch
import numpy as np
import matplotlib.pyplot as plt

def loss_func(w):
  return torch.sin(w[0])*torch.cos(w[1])

epochs=1000
lr=1e-1
results={}
```

# Оптимизатор SGD

```python
w=torch.tensor([10.0,10.0],requires_grad=True)

x,y,loss=[],[],[]
optimizer=torch.optim.SGD([w],lr=lr)

for step in range(epochs):
  function=loss_func(w)
  function.backward()
  optimizer.step()
  optimizer.zero_grad()

  x.append(w[0].item())
  y.append(w[1].item())
  loss.append(function.item())

results['SGD']=(x,y,loss)
```

# Оптимизатор LBFGS

```python
w = torch.tensor([10.0, 10.0], requires_grad=True)

x, y, loss = [], [], []
optimizer = torch.optim.LBFGS([w], lr=lr)

for step in range(epochs):
    def closure():
        optimizer.zero_grad()
        function = loss_func(w)
        function.backward()
        return function

    optimizer.step(closure)

    x.append(w[0].item())
    y.append(w[1].item())
    loss.append(loss_func(w).item())

results['LBFGS'] = (x, y, loss)
```

# Отрисовка

```python
fig, ax = plt.subplots(figsize=(10, 6))

for name, (px, py, ploss) in results.items():
    ax.plot(ploss, label=f'{name} (Loss: {ploss[-1]:.4f})')

ax.set_title('Изменение функции потерь по эпохам')
ax.set_xlabel('Эпохи')
ax.set_ylabel('Значение Loss')
ax.legend()
ax.grid(True)
```
