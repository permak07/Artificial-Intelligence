```python
import numpy as np
import gensim.downloader as api

model = api.load('word2vec-google-news-300')

similar = [
    ('Russia', 'Moscow'), ('France', 'Paris'), ('Germany', 'Berlin'), 
    ('Italy', 'Rome'), ('Spain', 'Madrid'), ('Japan', 'Tokyo'), 
    ('China', 'Beijing'), ('Egypt', 'Cairo'), ('India', 'Delhi'), 
    ('UK', 'London'), ('USA', 'Washington'), ('Canada', 'Ottawa'),
    ('Australia', 'Canberra'), ('Brazil', 'Brasilia'), ('Argentina', 'Buenos_Aires'),
    ('Belgium', 'Brussels'), ('Turkey', 'Ankara'), ('Greece', 'Athens'),
    ('Sweden', 'Stockholm'), ('Norway', 'Oslo'), ('Finland', 'Helsinki'),
    ('Denmark', 'Copenhagen'), ('Poland', 'Warsaw'), ('Austria', 'Vienna'),
    ('Hungary', 'Budapest'), ('Portugal', 'Lisbon'), ('Ireland', 'Dublin'),
    ('Thailand', 'Bangkok'), ('Vietnam', 'Hanoi'), ('Indonesia', 'Jakarta'),
    ('Malaysia', 'Kuala_Lumpur'), ('South_Korea', 'Seoul'), ('Iran', 'Tehran'),
    ('Iraq', 'Baghdad'), ('Syria', 'Damascus'), ('Lebanon', 'Beirut'),
    ('Israel', 'Jerusalem'), ('Jordan', 'Amman'), ('Saudi_Arabia', 'Riyadh'),
    ('Cuba', 'Havana'), ('Jamaica', 'Kingston'), ('Chile', 'Santiago'),
    ('Peru', 'Lima'), ('Colombia', 'Bogota'), ('Venezuela', 'Caracas'),
    ('Kenya', 'Nairobi'), ('South_Africa', 'Pretoria'), ('Morocco', 'Rabat'),
    ('Nigeria', 'Abuja'), ('Ghana', 'Accra')
]
print(f"{len(similar)} пар с одинаковым смысловым отношением")

different = [
    ('Russia', 'bear'), ('USA', 'Trump'), ('China', 'dragon'), 
    ('India', 'cow'), ('Australia', 'kangaroo'), ('Canada', 'maple'),
    ('France', 'wine'), ('Germany', 'beer'), ('Italy', 'pasta'),
    ('Japan', 'sushi'), ('Mexico', 'taco'), ('Brazil', 'carnival'),
    ('Spain', 'bull'), ('UK', 'tea'), ('Egypt', 'pyramid'),
    ('Greece', 'mythology'), ('Turkey', 'coffee'), ('Sweden', 'IKEA'),
    ('Norway', 'fjord'), ('Finland', 'sauna'), ('Denmark', 'Lego'),
    ('Poland', 'sausage'), ('Austria', 'Mozart'), ('Hungary', 'goulash'),
    ('Portugal', 'port'), ('Ireland', 'shamrock'), ('Thailand', 'boxing'),
    ('Vietnam', 'rice'), ('Indonesia', 'island'), ('Malaysia', 'towers'),
    ('South_Korea', 'Samsung'), ('Iran', 'carpet'), ('Iraq', 'oil'),
    ('Syria', 'war'), ('Lebanon', 'cedar'), ('Israel', 'tech'),
    ('Jordan', 'Petra'), ('Saudi_Arabia', 'desert'), ('Cuba', 'cigar'),
    ('Jamaica', 'reggae'), ('Chile', 'Andes'), ('Peru', 'Inca'),
    ('Colombia', 'emerald'), ('Venezuela', 'oil'), ('Kenya', 'safari'),
    ('South_Africa', 'gold'), ('Morocco', 'couscous'), ('Nigeria', 'cinema'),
    ('Ghana', 'cocoa'), ('Argentina', 'tango')
]
print(f"{len(different)} пар с разным смысловым отношением")
```

**Получение векторов отношений (разностей эмбеддингов)** 

```python
def word2vec(word1,word2,model):
    return model[word1]-model[word2]

sim=[]
for words in similar:
    emb=word2vec(words[0],words[1],model)
    if emb is not None:
        sim.append(emb)

dif=[]
for words in different:
    emb=word2vec(words[0],words[1],model)
    if emb is not None:
        dif.append(emb)
```

**Расчет попарных Евклидовых расстояний внутри групп**  

```python
def distance(value1, value2):
    return np.sqrt(np.sum((value1 - value2) ** 2))

sim_dist = []
for i in range(len(sim)):
    for j in range(i + 1, len(sim)):
        dist_pair = distance(sim[i], sim[j])
        sim_dist.append(dist_pair)

dif_dist = []
for i in range(len(dif)):
    for j in range(i + 1, len(dif)):
        dist_pair = distance(dif[i], dif[j])
        dif_dist.append(dist_pair)
```

**Результат**

```python
avg_sim_dist = np.mean(sim_dist)
avg_dif_dist = np.mean(dif_dist)

print(f'Среднее одинаковых отношений |(Russia-Moscow)-(USA-Washington)|: {avg_sim_dist:.4f}')
print(f'Среднее разных отношений |(Russia-Bear)-(USA-Trump)|: {avg_dif_dist:.4f}\n')
```
